## 1. 输入轻轻动一下

设 $f(x)=3x^2-4x+5$。导数衡量输入微小变化时，输出的变化率：

$$f'(x)=\lim_{h\to0}\frac{f(x+h)-f(x)}{h}=6x-4.$$

先预测：在 $x=3$ 处，增大一点 $x$，输出会增大还是减小？

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return 3*x**2 - 4*x + 5

x = 3.0
for h in [0.1, 0.01, 0.001, 0.0001]:
    print(f"h={h:g}, 数值导数={(f(x+h)-f(x))/h:.6f}")
print("解析导数:", 6*x - 4)
```

```python
xs = np.linspace(-2, 4, 200)
plt.plot(xs, f(xs), label="f(x)")
plt.plot(xs, f(x) + (6*x-4)*(xs-x), "--", label="tangent at x=3")
plt.scatter([x], [f(x)])
plt.legend()
plt.grid()
plt.show()
```

**动手改：** 把上一段的 $x$ 改为 $-1$ 和 $2/3$，分别预测导数，再重新运行。注意斜率接近零时，有限差分仍可能有小误差。

## 2. 多个输入：一次只改变一个

$$L=ab+c.$$

偏导数表示固定其他输入，只改变某一个输入时的变化率。先手算 $a=2,b=-3,c=10$ 时的 $L$，以及它对三个输入的偏导数。

```python
a, b, c = 2.0, -3.0, 10.0
h = 1e-5
L = a*b + c
print("L =", L)
print("dL/da ≈", ((a+h)*b+c-L)/h)
print("dL/db ≈", (a*(b+h)+c-L)/h)
print("dL/dc ≈", (a*b+(c+h)-L)/h)
```

## 3. 用 micrograd 验证

`.data` 保存前向计算的值，`.grad` 保存最终输出对该节点的导数。`backward()` 从输出开始沿计算图反向传播。

这里每次运行都新建计算图，避免旧梯度累积。先观察结果，后续课程再自己实现 `Value`。

```python
from micrograd.engine import Value

a, b, c = Value(2.0), Value(-3.0), Value(10.0)
L = a*b + c
L.backward()
print("L =", L.data)
print("梯度:", a.grad, b.grad, c.grad)
```
